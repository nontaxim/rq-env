# Exercise 1: First Run

```python
from pyquil import Program
from pyquil.gates import *
from pyquil import get_qc

program = Program()
program += X(0)
program += CNOT(0, 1)
readout = program.declare('readout', 'BIT', 2)
program += MEASURE(0, readout[0])
program += MEASURE(1, readout[1])
print(program)
```

```python
quantum_computer = get_qc('5q-qvm')
# print(quantum_computer)
executable = quantum_computer.compile(program)
result = quantum_computer.run(executable)
print(result.get_register_map()['readout'])
```

```python
program.wrap_in_numshots_loop(10)
executable = quantum_computer.compile(program)
result = quantum_computer.run(executable)
print(result.get_register_map()['readout'])
```

## "The output of X gate is similar to the output of Hadamard-Z-Hadamard gates."

### Build circuits to prove above sentence and discuss why both opeartions show the same result.

#### 1. if the initial state of qubit is |0⟩

##### X gate

```python
program_q1_x_0 = Program(X(0))
program_q1_x_0.measure_all()
program_q1_x_0.wrap_in_numshots_loop(10)
executable_q1_x_0 = quantum_computer.compile(program_q1_x_0)
result_q1_x_0 = quantum_computer.run(executable_q1_x_0)
print(result_q1_x_0.get_register_map()['ro'].T[0])
```

$$   
\begin{align*}
            X \cdot \ket{0}   &= \begin{bmatrix}
                    0 & 1 \\
                    1 & 0
                    \end{bmatrix}
                        \cdot
                    \begin{bmatrix}
                    1 \\
                    0
                    \end{bmatrix}\\
                &= \begin{bmatrix}
                    0 \\
                    1
                    \end{bmatrix} \\
                &= \ket{1}
\end{align*}
$$

##### H gate &#8594; Z gate &#8594; H gate

```python
program_q1_hzh_0 = Program(H(0), Z(0), H(0))
program_q1_hzh_0.measure_all()
program_q1_hzh_0.wrap_in_numshots_loop(10)
executable_q1_hzh_0 = quantum_computer.compile(program_q1_hzh_0)
result_q1_hzh_0 = quantum_computer.run(executable_q1_hzh_0)
print(result_q1_hzh_0.get_register_map()['ro'].T[0])
```

$$
\begin{align*}  
            H \cdot Z \cdot H \cdot \ket{0}   &= 
                    \frac{1}{\sqrt{2}} 
                        \cdot 
                    \begin{bmatrix}
                    1 & 1 \\
                    1 & -1
                    \end{bmatrix}
                        \cdot
                    \begin{bmatrix}
                    1 & 0 \\
                    0 & -1
                    \end{bmatrix}
                        \cdot
                    \frac{1}{\sqrt{2}} 
                        \cdot 
                    \begin{bmatrix}
                    1 & 1 \\
                    1 & -1
                    \end{bmatrix}
                        \cdot
                    \begin{bmatrix}
                    1 \\
                    0
                    \end{bmatrix}\\

                &= \frac{1}{2}
                        \cdot 
                    \begin{bmatrix}
                    1 & 1 \\
                    1 & -1
                    \end{bmatrix}
                        \cdot
                    \begin{bmatrix}
                    1 & 0 \\
                    0 & -1
                    \end{bmatrix}
                        \cdot
                    \begin{bmatrix}
                    1 \\
                    1
                    \end{bmatrix}
                        \\

                &= \frac{1}{2}
                        \cdot 
                    \begin{bmatrix}
                    1 & 1 \\
                    1 & -1
                    \end{bmatrix}
                        \cdot
                    \begin{bmatrix}
                    1 \\
                    -1
                    \end{bmatrix}
                        \\

                &= \frac{1}{2}
                        \cdot
                    \begin{bmatrix}
                    0 \\
                    2
                    \end{bmatrix}
                        \\

                &= \begin{bmatrix}
                    0 \\
                    1
                    \end{bmatrix}\\

                &= \ket{1}
\end{align*}
$$

We can conclude that
$$
\begin{align*}
H \cdot Z \cdot H \cdot \ket{0}   = X \cdot \ket{0}
\end{align*}
$$
